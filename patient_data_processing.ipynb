```python
# Patient Data Processing Pipeline
# Healthcare Analytics - Manual Step

import datetime

# Sample patient data (simulating what would come from a real database)
patients = [
    {"id": 1, "name": "Raj Kumar",    "age": 45, "diagnosis": "Diabetes",    "admitted": True},
    {"id": 2, "name": "Priya Singh",  "age": 32, "diagnosis": "Hypertension","admitted": False},
    {"id": 3, "name": "Anil Mehta",   "age": 60, "diagnosis": "Diabetes",    "admitted": True},
    {"id": 4, "name": "Sunita Rao",   "age": 28, "diagnosis": "Asthma",      "admitted": False},
    {"id": 5, "name": "Vikram Das",   "age": 55, "diagnosis": "Hypertension","admitted": True},
    {"id": 6, "name": "Meera Patel",  "age": 38, "diagnosis": "Diabetes",  "admitted": True},
    {"id": 7, "name": "Meera Bhai",  "age": 38, "diagnosis": "Diabetes",  "admitted": True},
]

# Analysis
total_patients = len(patients)
avg_age = sum(p["age"] for p in patients) / total_patients
admitted_count = sum(1 for p in patients if p["admitted"])

# Diagnosis breakdown
diagnosis_count = {}
for p in patients:
    d = p["diagnosis"]
    diagnosis_count[d] = diagnosis_count.get(d, 0) + 1

# Report
print("=" * 40)
print("  PATIENT DATA SUMMARY REPORT")
print(f"  Generated: {datetime.datetime.now()}")
print("=" * 40)
print(f"Total Patients    : {total_patients}")
print(f"Average Age       : {avg_age:.1f} years")
print(f"Currently Admitted: {admitted_count}")
print(f"\nDiagnosis Breakdown:")
for diagnosis, count in diagnosis_count.items():
    print(f"  {diagnosis}: {count} patients")
print("=" * 40)
```
